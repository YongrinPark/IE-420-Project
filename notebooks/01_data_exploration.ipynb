# 01 – Data Exploration

**Goal:** Download daily and intraday prices, inspect the data, compute log returns,
and visualise the basic statistical properties of the portfolio assets.

Run all cells top-to-bottom from the **project root**.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import config
from src.data_loader import download_daily_prices, download_intraday_prices
from src.preprocess  import compute_log_returns, align_price_data, handle_missing_values

print('Config loaded. Tickers:', config.TICKERS)
```

## 1. Download Daily Prices

```python
daily = download_daily_prices(save=True)
daily.tail()
```

## 2. Compute Log Returns

```python
log_returns = compute_log_returns(daily)
print(f'Log returns shape: {log_returns.shape}')
log_returns.describe().round(6)
```

## 3. Price History Plot

```python
fig, ax = plt.subplots(figsize=(12, 5))
normalised = daily / daily.iloc[0] * 100
normalised.plot(ax=ax)
ax.set_title('Normalised Price History (Base = 100)')
ax.set_ylabel('Index')
ax.legend(loc='upper left')
plt.tight_layout()
os.makedirs('../results/figures', exist_ok=True)
plt.savefig('../results/figures/normalised_prices.png', dpi=150)
plt.show()
```

## 4. Return Distribution

```python
fig, axes = plt.subplots(1, len(config.TICKERS), figsize=(15, 4))
for ax, ticker in zip(axes, config.TICKERS):
    log_returns[ticker].hist(bins=60, ax=ax, edgecolor='k', alpha=0.7)
    ax.set_title(ticker)
    ax.set_xlabel('Log Return')
plt.suptitle('Daily Log Return Distributions')
plt.tight_layout()
plt.savefig('../results/figures/return_distributions.png', dpi=150)
plt.show()
```
